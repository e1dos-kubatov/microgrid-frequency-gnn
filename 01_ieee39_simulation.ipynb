```python
import andes
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

t_event = 2.0   # moment of disturbance, seconds

def run_load_step(pq_number, factor=1.2, t_end=20):
    ss = andes.load(andes.get_case('ieee39/ieee39_full.xlsx'),
                    setup=False, no_output=True)
    pq_ids = ss.PQ.idx.v
    ss.add('Alter', dict(model='PQ', dev=pq_ids[pq_number], t=t_event,
                         src='Ppf', attr='v', method='*', amount=factor))
    ss.setup()
    ss.PFlow.run()
    ss.TDS.config.tf = t_end
    ok = ss.TDS.run()
    return ss, ok

for k in range(5):
    ss, ok = run_load_step(k, factor=1.2)
    print(f"Load {k}: {'stable, OK' if ok else 'unstable'}")
    if ok:
        break
```
